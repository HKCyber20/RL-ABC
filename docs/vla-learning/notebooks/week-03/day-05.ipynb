## Context：知识点及其在 VLA 中的作用

BC 最小化专家状态分布下的动作误差：

\[
\mathbb E_{s\sim d_{expert}}[\ell(\pi(s),a^*)]
\]

部署时状态来自 `d_pi`。微小动作偏差会将机器人带到示范未覆盖区域，之后预测更差，形成误差累积。这解释了为何验证 MSE 很低，VLA 仍可能抓空或越走越偏。


## Concepts：概念、公式、形状与数据流

一维系统 `x_{t+1}=x_t+a_t`，专家 `a_t=k(x*-x_t)`。训练数据只覆盖理想轨迹附近；给策略加入偏差 `b`：

\[
\hat a_t=a_t^*+b
\]

单步误差是 `|b|`，若没有反馈纠正，T 步位置偏差可随 T 累积。闭环策略可能纠正，但若分布外预测错误随偏离增大，仍会发散。

```text
expert states -> supervised fit
policy error -> new state outside expert support
             -> larger model error -> further drift
```


## Learning Steps

1. 画 expert/policy 两条状态轨迹。
2. 在同一 expert state 上测 one-step error。
3. 从相同初态 rollout 20 步，记录 state error。
4. 从略有扰动的初态重跑，比较恢复能力。
5. 列举增加数据覆盖、噪声增强、DAgger 类交互收集等缓解方式及成本。

## Help：排查、最低完成线与提高

### 常见错误

- rollout 每步重置到专家状态：这叫 teacher forcing，不是闭环。
- 将环境噪声和策略误差混在一起：先做无噪声对照。
- 只看最终点漏掉中途越界：记录最大偏离/安全事件。
- 失败后缩短 horizon 让结果好看：horizon 是评估协议的一部分。

### 最低完成线

完成一个训练内、一个训练外初态的 10 步对比，并同时报告一步 MSE 与最终误差。

### 可选提高

模拟一次 DAgger 迭代：在策略访问状态上查询专家标签，重新拟合后比较；注明现实机器人查询专家的成本与安全约束。
