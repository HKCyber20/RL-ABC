## Context：知识点及其在 VLA 中的作用

增强可缓解背景、亮度和小位移过拟合，但 VLA 的图像与动作存在几何对应。随意水平翻转或改变指令关键颜色，可能把正确动作变成错误标签。


## Concepts：概念、公式、形状与数据流

训练时 `I'=g(I;ξ)`，若变换不改变控制语义则标签保持 `a'=a`；若是几何等变变换，必须有一致的 `a'=T_g(a)`。例如以相机横轴为基准的水平镜像可能要求 `Δx'=-Δx`，但只有坐标约定明确时才成立。

```text
train: raw → stochastic augment → fixed normalize
val/test/rollout: raw → fixed normalize
```

3. 先启用轻微亮度/对比度和小裁剪；颜色是指令条件时禁用改变类别的 hue shift。
4. 固定 augment seed，保存同一帧的 8 个样本与参数。
5. 确认 val/test 不经过随机增强；对 batch 做 shape/finite 检查。

## Help：排查、最低完成线与提高

### 常见错误

| 错误 | 排查 |
|---|---|
| crop 把目标完全裁掉 | 加目标可见率检查或减小范围 |
| hue jitter 改变“红/蓝” | 颜色条件任务只做保色光度变换 |
| transform 顺序错误 | 先几何/光度，再 tensor normalize，并记录参数 |
| 每个 worker 随机性失控 | 设置 worker seed 并记录 generator |

### 最低完成线

使用 CPU 和 16 张 synthetic 任务图像，只实现小范围亮度和随机平移/裁剪中的一种安全方案，完成 6 类安全矩阵与 val 无随机增强检查。
