## Context：知识点及其在 VLA 中的作用

加入语言前必须确认视觉真的提供了目标信息。严格 baseline 能揭示 state 是否泄漏目标、图像是否被模型忽略，以及视觉编码成本是否带来闭环收益。


## Concepts：概念、公式、形状与数据流

公平对照保持：episode split、seed、训练步数、动作 loss、normalization 和 rollout seeds 相同，只改变视觉输入。比较表至少包含参数量、训练耗时、test MAE、夹爪准确率、成功数/总数。

模态依赖检查：

```text
正常图像
图像置零/打乱
正常 state
state 置零/打乱
```

若图像打乱后成功率不变，应优先怀疑数据泄漏或模型忽略视觉。


## Learning Steps

1. 冻结评估协议和两个模型配置，只保留必要差异。
2. 从头或载入可复现 checkpoint，收集同尺度曲线。
3. 对两模型执行同一组至少 10 个低资源或 20 个主路径 rollout。
4. 对视觉模型做图像置零/跨 episode 打乱测试。
5. 选 3 个失败 episode，按感知、状态、控制、数据或判定器分类。
6. 闭卷完成本周测验，并写下第 7 周加入语言前的接口冻结项。

## Help：排查、最低完成线与提高

### 常见错误

| 错误 | 排查 |
|---|---|
| state-only 也近乎完美 | 检查 state 是否包含 object/goal pose 泄漏 |
| visual 模型明显更差 | 先查预处理、encoder 梯度、数据量和训练步数 |
| 打乱图像仍不变 | 检查融合连通、目标泄漏和任务是否真的需要视觉 |
| 两模型不可比 | 固定 split、seed、loss、步数和评估 episodes |

### 最低完成线

用 CPU synthetic 数据训练两个 tiny 模型；各做 10 次 Mock rollout；完成图像打乱消融和 1 个失败案例。保留所有实际日志。

### 可选提高

增加 image-only baseline；比较 frozen 与 fine-tune；用配对 bootstrap 估计成功率差异。
