## Context：知识点及其在 VLA 中的作用

RT-2 把 VLM 的语义先验与机器人控制连接起来；它提醒我们评估的不只是见过任务拟合，还包括语义与物体关系的迁移。


## Concepts：概念、公式、形状与数据流

- 抽象目标：对混合数据最大化条件 token 的对数似然，机器人样本预测动作 token，VQA/描述样本预测文本 token；不同样本使用不同输出语义和 loss mask。
- 机器人动作序列仍需可逆映射 `token ↔ normalized action ↔ physical action`，语言词表中的编号不能脱离 action tokenizer 解释。
- “出现新物体”不等于组合泛化；训练/测试切分必须记录物体、指令模板、关系与场景是否重叠。
