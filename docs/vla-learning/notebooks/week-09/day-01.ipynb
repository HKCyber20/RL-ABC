## Context：知识点及其在 VLA 中的作用

RT-1 展示了把机器人策略扩展为大规模序列模型的基础范式；理解动作离散化能帮助你判断 token 策略的精度、吞吐与安全边界。


## Concepts：概念、公式、形状与数据流

- 抽象数据流：`RGB [B,T,3,H,W] + instruction [B,L] → visual/language tokens → policy → action bins [B,T,D_a]`。
- 每个连续动作维度可按区间 `[m_j,M_j]` 分成 `K` 个箱：`q_j=clip(round((a_j-m_j)/(M_j-m_j)*(K-1)),0,K-1)`；解码得到箱中心，存在量化误差。
- RT-1 的工程启示是高维视觉 token 压缩、语言条件化与动作 token 预测；论文结构细节要与作者原图核对，不把抽象图误当唯一实现。
