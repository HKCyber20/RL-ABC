## Context：知识点及其在 VLA 中的作用

通用 VLA 的能力很大程度来自多源数据；能审计异构性，才不会把“轨迹更多”误当作“监督一致”。


## Concepts：概念、公式、形状与数据流

- OXE 是多个机器人数据集的混合生态；课程不依赖具体数据集数量，因为版本和子集会变化，需以所用 manifest 为准。
- 一个 episode 常含 `steps`，每步可能有多相机 observation、语言任务、proprioception、原始 action 和边界标志。
- 混合前至少核对：本体、控制模式、坐标系、单位、频率、相机命名、图像编码、动作维度、终止定义、许可与来源。


## Learning Steps

1. 绘制两个虚拟本体 A/B 的原始字段表。
2. 标记哪些字段能直接合并、哪些需转换、哪些必须保留 dataset-specific mask。
3. 编写一页 dataset card，记录来源、任务、样本量口径、动作语义和已知限制。
4. 设计 mixture sampling：按轨迹数、按数据集均衡或温度采样，并解释取舍。
