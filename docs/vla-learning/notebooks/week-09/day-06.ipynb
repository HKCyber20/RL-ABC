## Context：知识点及其在 VLA 中的作用

真实 OXE 下载昂贵且环境复杂；mock inspector 让你先验证最危险的数据逻辑，再决定是否投入下载和 GPU 资源。


## Concepts：概念、公式、形状与数据流

- 检查层次：文件/解析 → episode 边界 → step 字段 → dtype/shape → finite/range → 时间对齐 → action contract → 汇总统计。
- 建议错误结构：`{episode_id, step_index, code, severity, message}`，便于统计而不是解析自由文本。
- shape 示例：image metadata `[3,64,64]`、state `[8]`、action `[7]`；实际像素可省略，用校验元数据完成 CPU 静态练习。

## Help：排查、最低完成线与提高

### 常见错误

- 用字符串 `"NaN"` 代替真正的非有限数测试却未说明。
- 检查器遇第一个错误就退出，遗漏批量问题。
- 默认补齐缺字段，掩盖坏数据。
- 若输出与预期不符，先打印 episode/step 索引、shape、dtype、单位和 dataset key，再检查模型。
- 不得把论文中的报告结果或本页“预期输出”写成本地实测。

实现 5 条规则，接受 1 个合法 episode、拒绝 2 个坏 episode，并附实际或明确标注为预期的输出。 最低完成线仍须提交证据，且只能获得“最低线通过”，后续需补齐完整产物。
