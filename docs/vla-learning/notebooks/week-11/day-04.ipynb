## Help：排查、最低完成线与提高

### 常见错误

- 指令模板泄漏目标标签。
- 训练中同义句已经包含 held-out 组合。
- 只评语言分类不评闭环成功。
- 先检查 manifest 行数、唯一 episode id、seed/condition 分布、缺失值和分母，再解释模型。
- 不得将 mock 记录、预期 CI 或未运行的 rollout 写成真实性能。

### 最低完成线

2×2 组合表、2 个 held-out 和 2 对反事实。 最低线仍需固定协议并提交可复核证据。

### 可选提高

加入同义改写与否定指令，单列分布外程度。
