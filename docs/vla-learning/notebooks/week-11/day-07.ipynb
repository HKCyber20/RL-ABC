## Context：知识点及其在 VLA 中的作用

消融用于检验“哪项设计造成改进”的假设；本周报告把结果、证据与限制收敛为可审阅结论。


## Concepts：概念、公式、形状与数据流

- 消融对照需冻结数据 split、seed、checkpoint 选择规则、episode 初态和评估脚本；只改变语言、状态输入、增强或动作 chunk 等一个因素。
- 报告结构：问题→协议→数据→指标→结果→失败→消融→限制→下一步；每个结论标 OBSERVED/INFERRED/HYPOTHESIS。
- 若只有 mock 日志，结论只能是分析管线正确性，不能报告策略泛化性能。

## Help：排查、最低完成线与提高

### 常见错误

- 选择最好的 seed 做展示。
- 结果相同就声称模块无用，忽略统计功效。
- 图表没有 n 或误差区间。
- 先检查 manifest 行数、唯一 episode id、seed/condition 分布、缺失值和分母，再解释模型。
- 不得将 mock 记录、预期 CI 或未运行的 rollout 写成真实性能。

### 最低完成线

完成周测、一个 2 条件消融、四切片表和 10 条失败分析。 最低线仍需固定协议并提交可复核证据。
