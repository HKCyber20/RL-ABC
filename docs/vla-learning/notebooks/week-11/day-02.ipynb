## Context：知识点及其在 VLA 中的作用

单次 rollout 的成功率波动很大；置信区间和 seed 明细让你判断差异是否稳定，而不是只比较两个百分点。


## Concepts：概念、公式、形状与数据流

- 对 k/n 的 Wilson 区间，令 z=1.96，中心与半宽由 `(p̂+z²/(2n) ± z*sqrt(p̂(1-p̂)/n+z²/(4n²)))/(1+z²/n)` 给出。
- 若多个 episode 共享 seed/场景，独立同分布假设可能不成立；同时报告 per-seed 成功率，并可按 seed bootstrap。
- 置信区间不是“真实成功率有 95% 概率在区间内”的贝叶斯陈述；本课关注可重复汇报而非过度解读显著性。
