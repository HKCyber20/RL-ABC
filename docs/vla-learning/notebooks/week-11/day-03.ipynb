## Context：知识点及其在 VLA 中的作用

未见位置测试策略是否学到视觉闭环与空间关系，而不是记忆固定抓取点；同时要防止不可达位置把控制限制误判成模型泛化失败。


## Concepts：概念、公式、形状与数据流

- 用 workspace 坐标 `(x,y)` 定义区域；训练/测试之间设置 buffer margin，避免连续空间边界近邻泄漏。
- 每个位置记录 reachability、visibility、初始末端距离、遮挡与目标大小；这些是重要混杂变量。
- 主指标按位置 cell 报成功率与 CI；附最差 cell、到训练集最近距离和失败阶段。


## Learning Steps

1. 画工作空间网格，标 train/validation/test 与 buffer。

3. 为每格平衡 seed、物体和指令。
4. 生成 per-cell 表或 ASCII 热图，分析边缘/遮挡效应。

## Help：排查、最低完成线与提高

### 常见错误

- 像素位置不同但世界坐标相同。
- 改变位置时同时改变背景。
- 只报平均而隐藏一个完全失败区域。
- 先检查 manifest 行数、唯一 episode id、seed/condition 分布、缺失值和分母，再解释模型。
- 不得将 mock 记录、预期 CI 或未运行的 rollout 写成真实性能。

### 最低完成线

4 个 train、4 个 test 点，margin 检查和 per-point 表。 最低线仍需固定协议并提交可复核证据。

### 可选提高

拟合成功率随最近训练距离的趋势，仅作探索性分析。
