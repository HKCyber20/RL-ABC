## Concepts：概念、公式、形状与数据流

- 配对设计保持初始物体/机器人状态、指令和 seed，改变一个 nuisance factor。
- 相对下降 `Δ=p_clean-p_perturbed`；同时报两个比例的 CI 和 paired outcome 表，不只给百分比。
- 干扰强度需可复现：亮度因子、纹理 id、distractor 数、相机位姿扰动范围；过强到不可见时应单列。

## Help：排查、最低完成线与提高

### 常见错误

- 随机背景同时随机位置。
- 把图像增强训练结果混入测试协议。
- 只展示成功案例截图。
- 先检查 manifest 行数、唯一 episode id、seed/condition 分布、缺失值和分母，再解释模型。
- 不得将 mock 记录、预期 CI 或未运行的 rollout 写成真实性能。

### 最低完成线

一种扰动、8 对记录、四格 paired outcome 表。 最低线仍需固定协议并提交可复核证据。

### 可选提高

做 saliency/attention 可视化，但明确其不等同因果解释。
