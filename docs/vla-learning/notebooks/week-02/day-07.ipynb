## Context：知识点及其在 VLA 中的作用

动作接口是模型与机器人之间的合同。它同时约束数据标签、模型输出、反归一化、坐标变换和控制执行；其中任一环节的约定不一致都会造成“模型数值看似正确、机器人行为错误”。


## Concepts：概念、公式、形状与数据流

周项目数据流：

```text
normalized prediction [B,7]
 -> denormalize/clip
 -> interpret in declared frame/unit
 -> compose T / validate rotation
 -> optional IK
 -> execute at declared Hz
 -> log new state and status
```

验证器至少检查：shape/dtype/finite、范围、frame、单位、旋转合法性、夹爪集合、频率、时间戳单调性。

## Help：排查、最低完成线与提高

### 常见错误

- 规范与代码常量重复且不一致：验证器从同一规范读取。
- 对非法值全都返回 `False` 无原因：输出字段级错误。
- clip 后声称原动作合法：同时记录原值与处理值。
- round trip 只测原点：至少再测非零点。

### 最低完成线

完成周测、完整动作规格、一个合法和两个非法验证、一个变换 round trip；其余登记为补救，整周暂不勾选。
