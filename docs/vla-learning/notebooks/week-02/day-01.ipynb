## Context：知识点及其在 VLA 中的作用

VLA 输出的 `Δx=0.02` 若不声明参考系就无法执行：在基座 x 轴、相机 x 轴或末端自身 x 轴上可能是三个不同方向。数据混用坐标系会让同一视觉情景对应冲突标签，模型 loss 难降或闭环方向错误。


## Concepts：概念、公式、形状与数据流

用 `{A}` 表示坐标系 A，`p^A` 表示点 p 在 A 系下的坐标。若 A 系中 B 系原点位置为 `t^A_B`，B 的朝向在 A 中为 `R^A_B`：

\[
p^A=R^A_B p^B+t^A_B
\]

形状为 `R [3,3]`、`p,t [3]`，批量时可为 `[B,3,3]` 和 `[B,3]`。常见坐标系：world、robot base、camera、end-effector/tool、object。命名必须同时表达“从谁到谁”。

## Help：排查、最低完成线与提高

### 常见错误

- 把坐标数值当物理点本身：同一点换基后数值会变。
- `R p + t` 写成 `R(p+t)`：平移所在坐标系不同。
- 图上只写 x/y 不画正向：无法检查符号。
- 默认相机 z 轴方向：查目标系统约定；本课练习需自行明确约定。
