## Context：知识点及其在 VLA 中的作用

VLA 可以输出关节命令，也可输出末端笛卡尔目标。后一种通常需 IK 转换给低层控制器。模型即使预测了合理末端目标，也可能因超出工作空间、关节限位或 IK 分支跳变而执行失败，因此必须区分策略失败与控制/求解失败。


## Concepts：概念、公式、形状与数据流

二维二连杆长度 `l1,l2`、角度 `q1,q2`：

$$
x=l_1\cos q_1+l_2\cos(q_1+q_2)
$$
$$
y=l_1\sin q_1+l_2\sin(q_1+q_2)
$$

IK 中：

$$
\cos q_2=\frac{x^2+y^2-l_1^2-l_2^2}{2l_1l_2}
$$

`q2` 通常有正负两支（肘上/肘下）。可达条件 `|l1-l2| <= sqrt(x²+y²) <= l1+l2`。一般 6/7 轴机械臂还涉及冗余、碰撞和关节限位。

## Help：排查、最低完成线与提高

### 常见错误

- 角度度/弧度混用：打印已知 `q=0` 结果。
- 浮点误差令 `acos` 略超 [-1,1]：先判断可达，再在容差范围 clip。
- 只比关节角：IK 多解时应比较 FK 末端误差。
- 将 IK 失败算作感知失败：日志需单独分类。

### 最低完成线

实现 FK，手算 `[1,1]` 的 IK 并回代；说明 `[2.1,0]` 为何不可达。
