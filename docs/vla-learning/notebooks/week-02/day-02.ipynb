## Context：知识点及其在 VLA 中的作用

机器人观测常来自相机，而控制在基座或末端坐标系执行。齐次变换把多段外参和末端位姿串联起来，是将视觉目标位置转换为动作目标的基本工具；方向写反会让策略向相反位置运动。


## Concepts：概念、公式、形状与数据流

\[
T^A_B=\begin{bmatrix}R^A_B&t^A_B\\0&1\end{bmatrix},\quad
\bar p^A=T^A_B\bar p^B
\]

链式组合：

\[
T^A_C=T^A_B T^B_C
\]

求逆：

\[
(T^A_B)^{-1}=T^B_A=
\begin{bmatrix}R^T&-R^Tt\\0&1\end{bmatrix}
\]

批量变换可为 `[B,4,4] @ [B,4,1] -> [B,4,1]`。矩阵乘法顺序从右向左作用。

## Help：排查、最低完成线与提高

### 常见错误

- 行向量/列向量混用：课程公式采用列向量；代码若采用行向量需明确转置约定。
- 忘加齐次 1：位置用 1，方向向量用 0。
- 组合次序颠倒：检查相邻上下标是否能“消去”。
- 用一般逆矩阵掩盖公式错误：同时实现解析逆并对比。
