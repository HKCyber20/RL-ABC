## Context：知识点及其在 VLA 中的作用

同一个“每步移动 1 cm”的策略在 5 Hz 与 20 Hz 下每秒位移相差四倍。模型推理慢于控制周期会导致动作重复、丢帧或使用陈旧观测。VLA 的闭环性能必须报告模型频率、环境频率、action horizon 和延迟策略。


## Concepts：概念、公式、形状与数据流

控制频率 `f`、周期 `dt=1/f`。若模型输出速度 `v`，位置增量应为 `Δx=v dt`；若模型直接输出“每步增量”，改变 f 会改变单位时间运动量。

```text
observe at t -> inference latency tau -> execute/hold action
             -> plant evolves -> observe at t+dt
```

一维闭环比例控制示例：

\[
e_t=x^*-x_t,\quad u_t=clip(k e_t),\quad x_{t+1}=x_t+u_t\Delta t+w_t
\]

开环序列不使用新的 `x_t`；闭环每步重新计算。


## Learning Steps

1. 分别计算 5/10/20 Hz 下 `dt`。
2. 比较速度动作与每步增量动作的单位。
3. 实现一维系统，加入小扰动 `w_t`。
4. 比较开环固定动作与闭环反馈误差曲线。
5. 加入 0、1、3 步观测延迟，记录成功和振荡。

## Help：排查、最低完成线与提高

### 常见错误

- 状态更新漏乘 dt：单位检查可发现。
- 动作与状态时间索引混一帧：画 `x_t,u_t,x_{t+1}`。
- 延迟实现实际读到未来：保存状态历史并显式索引。
- 每组用不同随机扰动：无法公平比较，需同 seed/同序列。

### 最低完成线

比较 5 Hz 与 20 Hz 的单位效果；完成无延迟和 1 步延迟两组 1D 闭环。
