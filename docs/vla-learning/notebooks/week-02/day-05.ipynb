## Context：知识点及其在 VLA 中的作用

“动作 7 维”本身信息不足。动作可为关节位置/速度/力矩，或末端绝对位姿/增量位姿；每种动作对数据分布、IK、稳定性和泛化有不同影响。VLA 数据跨机器人使用前，必须对动作语义和范围进行映射。


## Concepts：概念、公式、形状与数据流

本课程后续基线采用：

```text
a_t = [dx,dy,dz, drx,dry,drz, g]  # [B,7]
```

其中平移增量单位 m、旋转向量增量单位 rad、均在明确参考系表达；`g` 的开合约定必须写清。对每维上下界 `low,high [7]`，映射到 `[-1,1]`：

\[
a_{norm}=2\frac{a-low}{high-low}-1
\]

反归一化后再做安全裁剪。绝对动作依赖全局标定；增量动作局部、数据范围较稳定，但误差会累积。

- 对所有维使用同一尺度：旋转与平移量纲不同。

- 训练归一化与执行反归一化范围不同：保存同一统计/规格版本。
- clip 发生但无日志：调试阶段至少统计裁剪比例。
