## Context：知识点及其在 VLA 中的作用

姿态标签最常见的灾难不是模型结构，而是约定不一致：度/弧度混用、XYZ/ZYX 不同、四元数 `xyzw/wxyz` 颠倒、符号双覆盖未处理。VLA 输出旋转增量前必须固定表示并在预处理/后处理两端一致。


## Concepts：概念、公式、形状与数据流

合法旋转矩阵满足：

\[
R^TR=I,\qquad \det(R)=1
\]

单位四元数 `q=[w,x,y,z]` 满足 `||q||=1`，且 `q` 与 `-q` 表示同一旋转。轴角以单位轴 `u` 和角度 `theta` 表示；旋转向量常写作 `r=theta u [3]`。欧拉角 `[3]` 紧凑但依赖顺序且可能万向节锁。旋转矩阵 `[3,3]` 稳健但冗余；四元数 `[4]` 适合组合/插值但需归一化。

## Help：排查、最低完成线与提高

### 常见错误

- 90 直接传入要求弧度的函数：先用 `pi/2`。
- q 分量顺序不一致：所有库调用旁写出转换。
- 只检查 `R^TR=I`：反射矩阵还需 `det=1`。
- 对增量欧拉角直接长期累加：可能产生奇异和漂移。

### 最低完成线

写清四元数与欧拉角约定，验证一个 90° R 和单位 q，并说明 q/-q 问题。
