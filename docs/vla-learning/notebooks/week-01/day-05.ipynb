## Context：知识点及其在 VLA 中的作用

Cross-attention 的 Q 与 K/V 来自不同序列。令文本为 Q、图像 patch 为 K/V，含义是“每个文本位置从图像区域取回信息”；令动作 query 为 Q，则含义是“每个待预测动作槽读取多模态上下文”。方向决定输出序列长度和每个输出位置代表什么。


## Concepts：概念、公式、形状与数据流

文本读取图像：

```text
text [B,L,D]  -> Q [B,H,L,Dh]
image [B,P,D] -> K,V [B,H,P,Dh]
weights [B,H,L,P]
output  [B,L,D]
```

例如 `B=2,H=4,L=6,P=16,D=32,Dh=8`，权重为 `[2,4,6,16]`。交换方向后，图像为 Q，权重变成 `[2,4,16,6]`，输出长度也变为 16。

Cross-attention 只完成条件读取；机器人状态可经 MLP 变为状态 token，或在池化后拼接。两者都是设计选择，必须写明。

## Help：排查、最低完成线与提高

### 常见错误

- Q/K 特征维不同：先分别投影到相同 `D_h`。
- K/V patch 次序不一致：两者必须对应同一视觉位置。
- 只交换标签未交换 query 数值：记录实际张量。
- 把高权重直接称作可解释性结论：还需干预、梯度或消融证据。

### 最低完成线

写对两种方向的形状，手算一个 query 对四个 patch 的权重，并说明输出长度跟随 Q。

### 可选提高

把机器人状态投影成一个 token，比较“加入 K/V 序列”和“池化后拼接”两种融合方式的优缺点。
