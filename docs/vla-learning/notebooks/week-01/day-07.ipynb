## Context：知识点及其在 VLA 中的作用

综合链路是：原始输入 token 化/投影并加入位置，Q/K/V 计算模态内或模态间读取，融合表示结合机器人状态预测动作，动作进入环境形成下一观测。周验收关注能否同时解释语义、形状、梯度和闭环，而非背单个定义。


## Concepts：概念、公式、形状与数据流

周项目固定接口：

```text
image [B,16,32]
text  [B, 6,32] -> text queries image -> fused [B,6,32]
state [B,8]                         -> prediction [B,7]
prediction + target -> MSE -> backward
```

注意力核心：`softmax(QK^T/sqrt(d_k), dim=-1)V`；闭环核心：`(o_t,l,s_t)->a_t->env->(o_{t+1},s_{t+1})`。
