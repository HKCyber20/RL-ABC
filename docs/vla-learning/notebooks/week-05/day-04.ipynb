## Context：知识点及其在 VLA 中的作用

视觉策略依赖像素与空间位置之间的稳定关系。相机翻转、外参方向写反或图像比动作晚一帧，都会制造难以从 loss 发现的系统性标签噪声。


## Concepts：概念、公式、形状与数据流

针孔投影：

$$
\tilde p = K[R\mid t]P_w,\quad K=\begin{bmatrix}f_x&0&c_x\\0&f_y&c_y\\0&0&1\end{bmatrix}.
$$

必须写明外参是 `T_camera_world` 还是 `T_world_camera`。RGB 原始 shape 通常 `[H,W,3] uint8`；多相机 batch 可为 `[B,V,3,H,W]`。采集时记录 `image_timestamp` 和 `action_timestamp`，并检查 `|Δt|`。


## Learning Steps

1. 选择主相机视角，说明第三人称与腕部相机各自覆盖和遮挡风险。
2. 保存分辨率、FOV 或 K、外参方向、颜色通道、上下方向、near/far。
3. 在场景中放三个已知位置标记点，投影或人工确认左右/上下/深度顺序。
4. 连续采集 20 帧，检查时间戳单调和相邻帧差异。
5. 将相机帧与最近控制时刻配对，输出最大、平均 `|Δt|`。

## Help：排查、最低完成线与提高

### 常见错误

| 错误 | 排查 |
|---|---|
| RGB 显示成蓝色 | 在读取与可视化边界检查通道顺序 |
| 外参投影镜像 | 验证变换方向、右手系和矩阵乘法顺序 |
| 腕部相机图像跳变 | 检查相机是否正确绑定末端 link |
| 图像/动作错一帧 | 同时记录采样前后时间戳，不用数组下标猜同步 |

### 最低完成线

使用 20 张 `64×64` synthetic RGB，完成元数据、方向标记、时间单调、非黑帧和同步误差五项检查。
