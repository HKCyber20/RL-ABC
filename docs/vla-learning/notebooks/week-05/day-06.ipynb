## Context：知识点及其在 VLA 中的作用

模仿学习的上限由示范数据决定。VLA 需要同一时刻的图像、指令、机器人状态和随后执行的动作；采集循环中的一帧偏移会把正确专家变成错误监督。


## Concepts：概念、公式、形状与数据流

单条轨迹：

\[
\tau=\{(o_t,l,s_t,a_t,r_t,o_{t+1},d_t,u_t)\}_{t=0}^{T-1}.
\]

正确循环：

```text
obs_t → expert(obs_t, instruction) → action_t
  └── record pre-action fields          ↓ env.step
record reward_t, next_obs, flags ← obs_{t+1}
```

采集频率 `f` 与控制周期 `Δt=1/f` 要固定或逐步记录。人工演示、脚本专家和 planner 都应记录 `expert_type`。


## Learning Steps

1. 定义 episode 开始元数据：seed、task、instruction、schema_version、expert_type。
2. 每步先冻结/复制 `obs_t`，再求动作，避免引用被环境原地修改。
3. 校验动作后执行 step，把 transition 一次性追加。
4. 结束后运行成功判定，失败轨迹也保存并注明原因。
5. 采集真实平台至少 20 条成功轨迹；低资源路径采集 5 条确定性 Mock，其中含成功与失败。
6. 汇总长度、成功率、动作越界数、时间间隔和阶段覆盖。
