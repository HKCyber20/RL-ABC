## Context：知识点及其在 VLA 中的作用

仿真平台决定传感器、动作频率、坐标约定、任务判定和数据格式。选择错误不是“以后换环境”这么简单，它会渗透到训练标签与评估协议。Mock 的意义是先固定契约，让后续数据和模型工作不被 GPU 或安装阻塞。


## Concepts：概念、公式、形状与数据流

平台决策至少比较：操作任务覆盖、API 稳定性、渲染后端、GPU/驱动要求、已有专家数据、许可证和复现成本。

统一环境契约：

```text
seed + task_config
      ↓
env.reset() → obs_0, info_0
      ↓ action_t
env.step(action_t) → obs_{t+1}, reward_t, terminated, truncated, info
```

低资源 Mock 可用二维状态 `s_t=[gripper_x, gripper_y, object_x, object_y, closed]`，动作 `a_t=[Δx, Δy, gripper_cmd]`，形状分别为 `[5]` 与 `[3]`。状态转移可写为 `p_{t+1}=clip(p_t+Δp, low, high)`。
