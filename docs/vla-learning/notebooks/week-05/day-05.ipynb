## Context：知识点及其在 VLA 中的作用

训练 loss 衡量动作拟合，成功判定才衡量任务。含糊或有漏洞的 success 会直接污染数据筛选、rollout 指标和失败分析。


## Concepts：概念、公式、形状与数据流

设物体位置 `p_o`、目标中心 `p_g`、目标容差 `ε`，基础空间条件为：

\[
c_{inside}=\|p_o-p_g\|_2<\epsilon.
\]

稳定成功可定义为 `inside ∧ released ∧ stable` 连续满足 `K` 帧；抓取可结合双指接触、夹爪闭合与物体随末端移动。阶段状态机：

```text
approach → grasp → lift → transport → release → stable_success
```

## Help：排查、最低完成线与提高

### 常见错误

| 错误 | 排查 |
|---|---|
| 物体途经目标即成功 | 要求释放并连续稳定 K 帧 |
| 接触等同抓住 | 检查闭合、双侧接触和随动 |
| 阈值无单位 | 名称与表格同时写 m/rad/frame |
| 只存最终 success | 同时存阶段首次达到时间，便于定位失败 |

### 最低完成线

在二维 Mock 上实现 inside、released、稳定 3 帧，并通过正常、掠过、未松爪三种序列。
