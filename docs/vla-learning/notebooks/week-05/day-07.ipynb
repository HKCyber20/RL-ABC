## Context：知识点及其在 VLA 中的作用

“文件能打开”不等于数据可训练。可回放性证明动作、状态转移、终止与成功标签自洽；manifest 让未来的训练、切分和复现实验能追溯到同一批数据。


## Concepts：概念、公式、形状与数据流

建议目录：

```text
dataset/
├── manifest.json
├── schema.json
└── episodes/episode_000001.npz
```

manifest 至少含 `dataset_id/schema_version/platform/task/episode_count/success_count/action_spec/camera_spec/created_by`。确定性 Mock 可验证：

\[
\max_t\|\hat s_{t+1}-s_{t+1}\|_\infty < \epsilon.
\]

真实物理仿真若非完全确定，应比较任务阶段、末端轨迹误差和最终 success，并记录容差。

## Help：排查、最低完成线与提高

### 常见错误

| 错误 | 排查 |
|---|---|
| 文件顺序被字典/文件系统改变 | 用显式 episode_id 与 step_index 排序 |
| replay 初始即不一致 | 检查 seed、环境版本、reset options 和随机源 |
| 压缩后 dtype 变化 | 加载后按 schema 逐字段断言 |
| 只比较最终状态 | 报告首个分叉步，避免误差抵消 |

### 最低完成线

交付 5 条 CPU Mock 轨迹、manifest/schema、成功与失败各一条逐步回放，以及一个动作篡改负向测试。
