## Context：知识点及其在 VLA 中的作用

VLA 对预处理漂移非常敏感：颜色通道、裁剪、指令模板或相机顺序改变，都可能让接口“能运行但行为错误”。


## Concepts：概念、公式、形状与数据流

- 原始图像通常是 uint8 [H,W,3] 或 PIL RGB；processor 输出形态依版本而异，验收应读取实际键/shape，而不是硬编码想象。
- prompt 模板属于训练分布的一部分；大小写、前后缀和空白应由一个函数生成并测试。
- 多相机输入要固定顺序与 mask；若所用模型只支持单图，不能简单在通道维拼接后声称兼容。


## Learning Steps

3. 制作 train/eval preprocessing diff 表。

## Help：排查、最低完成线与提高

### 常见错误

- 把 resize 后 shape 当作原始图像证据。
- float 图像 [0,255] 被再次乘 255。
- 相机排序依赖字典偶然顺序。
- 先打印输入 mode/shape、processor keys、model revision、action stats key 与 action shape，再排查数值。
- 不得声称执行过未实际运行的安装、推理或训练。

### 最低完成线
