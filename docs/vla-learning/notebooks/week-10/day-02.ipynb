## Context：知识点及其在 VLA 中的作用

7B 级 VLA 的失败常发生在模型运行前；资源规划避免长下载、依赖冲突和显存不足，并保证没有高端 GPU 也能继续学习。


## Concepts：概念、公式、形状与数据流

- 粗略权重内存下界：参数量 × 每参数字节；训练还需梯度、优化器状态、激活和通信缓冲，不能用权重大小代表总显存。
- 环境记录至少含 OS、Python、PyTorch/CUDA、transformers、tokenizers、timm、flash-attn/替代、GPU/VRAM、磁盘、commit 和模型 revision。
- 三档路径：真实 checkpoint 推理；小 batch/adapter dry-run；纯 CPU mock 静态与接口测试。安装/下载是有成本写操作，课程不自动执行。
