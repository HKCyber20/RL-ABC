## Context：知识点及其在 VLA 中的作用

同一位置可能对应“正接近目标”和“正远离目标”两种速度，单帧位置无法区分。图像中的运动、夹爪接触和动作阶段也常需历史。历史窗口使策略近似恢复隐藏状态，但增加延迟、内存和时间对齐复杂度。


## Concepts：概念、公式、形状与数据流

长度 `H_o` 的历史：

\[
h_t=(x_{t-H_o+1},\ldots,x_t)
\]

低维形状 `[B,H_o,d_s]`；图像可为 `[B,H_o,C,H,W]` 或编码后 `[B,H_o,D]`。episode 开头可：重复首帧、零 padding + valid mask、或丢弃不足窗口的样本。三种策略不可混用且需记录。

窗口索引必须限制在同一 episode：`start=max(0,t-H_o+1)`，padding 只补当前 episode。


## Learning Steps

1. 构造两条在 `t` 位置相同但速度相反的轨迹。

3. 实现 `get_history(episode,t,H_o)`。
4. 输出历史 `[H_o,d]` 和 valid `[H_o]`。
5. 比较 repeat-first 与 zero+mask 的模型可见信息。

## Help：排查、最低完成线与提高

### 常见错误

- 窗口写成未来 `[t:t+H]`：部署不可用，检查最后索引应为 t。
- repeat-first 却还标 mask=False：定义需一致。
- shuffle 原始帧后再构窗：先按 episode/time 建窗，再 shuffle 样本。
- 图像历史直接沿通道拼却未记录：形状与预训练编码器接口可能不兼容。

### 最低完成线
