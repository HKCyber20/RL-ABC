## Context：知识点及其在 VLA 中的作用

Grounding 是把词语绑定到可见实体与控制目标。训练 loss、注意力热图和“偶尔成功”都不足以证明绑定；反事实测试能直接询问：只改变语义，决策是否按预期改变？


## Concepts：概念、公式、形状与数据流

控制变量矩阵：

```text
固定场景，换指令        → 应换目标
固定指令，交换物体位置  → 动作方向应跟随目标位置
删除关键颜色 token      → 置信/行为应退化或变得不确定
换无关措辞              → 语义等价时行为应近似稳定
打乱图像—文本配对       → 成功率应下降
```

首步方向一致率可定义为：

$$
DAcc=\frac1N\sum_i \mathbb{1}[\langle \hat{\Delta p}_i,p^*_{i}-p^{ee}_i\rangle>0].
$$


## Learning Steps

1. 生成至少 8 个双目标场景，目标左右位置随机而颜色身份独立。
2. 每个场景生成指向不同目标的成对指令，共至少 16 个样本。
3. 计算预测动作、目标距离变化或首步方向一致率。
4. 交换目标位置但保留指令；删除/替换关键 token；加入等价模板。
5. 若有 attention，记录热图，但以动作/目标选择作为主要证据。
6. 把失败分为未用语言、未定位视觉、错绑定、动作执行异常。

## Help：排查、最低完成线与提高

### 常见错误

| 错误 | 排查 |
|---|---|
| 颜色与左右固定绑定 | 每种颜色均出现在左右两侧 |
| 首步很小导致符号不稳 | 同时报投影值/范数并设最小动作阈值 |
| 等价模板含 OOV | 先区分 tokenizer OOV 与语义鲁棒性 |
| 打乱配对仍成功 | 检查任务是否可忽略语言或 success 判定过松 |
