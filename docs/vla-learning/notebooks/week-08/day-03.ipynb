## Context：知识点及其在 VLA 中的作用

RT-2 等工作把动作表示成 token，使语言模型可统一生成文本式序列。代价是量化误差、序列长度、跨维依赖和动作边界设计。本日只做可验证编码器，不要求训练大语言模型。


## Concepts：概念、公式、形状与数据流

对每一动作维在 `[l_j,u_j]` 分为 `K` 个 bin：

\[
q_j=clip\left(\left\lfloor\frac{a_j-l_j}{u_j-l_j}K\right\rfloor,0,K-1\right),
\]

反量化取 bin 中心：

\[
\tilde a_j=l_j+(q_j+0.5)\frac{u_j-l_j}{K}.
\]

单维最大中心量化误差约为 bin 宽一半。token 可按“每维一个 token”或联合 codebook；本课使用每维方案，输入 `[B,H,d_a]`，token 同 shape int64。


## Learning Steps

1. 只用训练集或物理安全边界定义每维 `[l,u]`，记录来源。
2. 实现 `encode_action`、`decode_tokens` 和越界计数，不静默吞掉越界。
3. 测边界、中心、随机动作、越界、NaN 五类输入。
4. 在 test 动作上报告逐维 MAE、最大误差、token 占用率和饱和率。
5. 比较 `K=16/64/256` 的误差与词表/序列成本。
6. 可选实现小分类 head `[B,H,d_a,K]` 并做交叉熵 forward/backward。

## Help：排查、最低完成线与提高

### 常见错误

| 错误 | 排查 |
|---|---|
| `a=u` 得到 token K | 公式后显式 clip 到 K-1，并测试上下边界 |
| 各维共用不合理范围 | 每维独立边界并写物理单位 |
| token id 与文本词表冲突 | 使用固定 offset/命名空间并保存 metadata |
| K 大但多数 bin 空 | 报占用率，考虑分位数 bin 或更多数据 |

### 最低完成线

CPU 随机动作上实现 K=16/64 的编解码、边界/NaN/越界测试与误差上界验证，不训练分类 policy。
