## Context：知识点及其在 VLA 中的作用

端到端失败可能来自感知、语言、融合、动作、控制或评估。直接换模型会把系统错误掩盖为“模型能力不足”；故障注入能证明你的检查真正有效。


## Concepts：概念、公式、形状与数据流

建议按可观测边界调试：

```text
1 数据/schema/时间
2 单模态预处理与编码
3 跨模态配对/grounding
4 normalized action 与 loss
5 denormalize/坐标/安全层
6 env transition/control
7 success/metrics/reporting
```

每层记录：假设、最小输入、预期不变量、观测证据、结论。不要以“训练 loss 下降”跳过前五层。


## Learning Steps

1. 为七层各列至少两个自动检查和一个人工可视化/抽查。
2. 注入 `action_{t+1}` 错位，观察时间对齐/回放检查是否报错。
3. 注入 RGB↔BGR 或图像翻转，观察预处理/探针指标。
4. 跨 episode 打乱文本，观察 paired grounding 检查与成功率。
5. 将动作反归一化重复一次或漏一次，观察范围/首步轨迹。
6. 放宽 success 条件制造假成功，确认阶段谓词/负向序列抓住。
7. 对每个故障记录“首个能检测到的层”，避免依靠最终失败猜测。

## Help：排查、最低完成线与提高

### 常见错误

| 错误 | 排查 |
|---|---|
| 同时注入多个故障 | 每次只改一个变量并保留 control |
| 注入后忘记关闭 | 配置显式标记并在正式评估拒绝 debug 模式 |
| 只看均值掩盖首帧异常 | 保存逐步日志和首个越界时间 |
| 成功判定器自己有 bug | 用 Day 5 合成序列先单测 checker |

### 最低完成线

CPU Mock 完成时间错位、文本打乱、动作重复反归一化三类注入，以及七层简化诊断矩阵。

### 可选提高

把检查整合为 preflight 命令；生成 failure bundle；增加相机外参和控制延迟注入。
