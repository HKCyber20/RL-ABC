## Context：知识点及其在 VLA 中的作用

本日证明最小 VLA 已形成闭环工程基线。这里先只做 ID，避免把系统未跑通与泛化能力混为一谈；第 11 周再做大规模泛化与失败分析。


## Concepts：概念、公式、形状与数据流

ID 条件应与训练任务保持同一属性集合、位置范围、相机配置和动作定义，但使用未训练的 episode seeds。报告：

\[
SR=\frac{N_{success}}{N_{rollout}},
\]

以及 approach/grasp/lift/place 阶段到达率、平均/中位 episode 长度、动作拒绝/clip 数、推理延迟、终止原因。正式评估必须冻结 checkpoint 和 seed 清单，失败不能删除重跑替换。

## Help：排查、最低完成线与提高

### 常见错误

| 错误 | 排查 |
|---|---|
| 只保存最终 success | 必须保留逐步动作、阶段和终止原因 |
| 失败后换 seed 不计入 | seed 清单预先冻结，所有尝试进入分母 |
| ID 与 train episode 重复 | 按 episode/scene seed 审计，而非只看文本 |
| 复跑结果不同 | 区分环境随机性、模型模式、预处理和控制时间差 |

### 最低完成线

CPU Mock 完成 10 次冻结 ID rollout、逐 episode JSON/CSV、阶段/安全统计、全部实际失败和一次新进程复跑；不得只运行成功样例。

### 可选提高

扩展到 30–50 次并计算区间；比较 `(H,K)`；录制成功/失败回放；尝试离散 token policy 但不替换连续主线结果。
